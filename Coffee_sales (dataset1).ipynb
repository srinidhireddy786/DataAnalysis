### 2. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
```

### 3. Load Dataset

```python
df = sns.load_dataset("Coffe_sales (4).csv")
df.head()
```

### 4. Explore Data

```python
print("Shape:", df.shape)
print(df.info())
print(df.isnull().sum())
print(df.describe())
print(df.columns.tolist())
```

### 5. Data Cleaning

```python
for col in ['Quantity', 'Sales', 'Unit Price']:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].mean())
for col in ['Product', 'Region', 'Customer']:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].mode()[0])

if 'Date' in df.columns:
    df = df.dropna(subset=['Date'])

print("Missing values after cleaning:\n", df.isnull().sum())
```

### Insights:
 
1.Missing values were filled or removed.

2.Duplicates were dropped.

3.Data was ready for visualization and statistical analysis.

### 6. Analysis with NumPy & Pandas

```python
# 1. Average Sale Amount
avg_sales = np.mean(df['Sales'])
print("Average Sale Amount:", round(avg_sales, 2))
 
# 2. Average Quantity Sold
avg_quantity = np.mean(df['Quantity'])
print("Average Quantity Sold:", round(avg_quantity, 2))

# 3. Total Revenue
total_revenue = df['Sales'].sum()
print("Total Revenue:", round(total_revenue, 2))
 
# 4. Average Sales by Product Type
sales_by_product = df.groupby('Product')['Sales'].mean().sort_values(ascending=False)
print("\nAverage Sales by Product:\n", sales_by_product)

```

### Insights:
 
1.Average age: ~29 years.
 
2.Overall survival: ~38% passengers survived.
 
3.By class: 1st class had the highest survival, 3rd class the lowest.
 
4.By sex: Females had higher survival than males.

### 7. Visualization with Matplotlib

```python

# Bar Chart: Sales by Coffee Type
plt.figure(figsize=(8,5))
df.groupby('Coffee_Type')['Sales'].sum().plot(kind='bar', color='peru')
plt.title('Total Sales by Coffee Type')
plt.xlabel('Coffee Type')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

#### Insights:
 
1.Visual confirms 2nd class passengers survived more.
 
2.Helps understand social/economic impact on survival.

```python
# Group by coffee type and calculate average sales
avg_sales = df.groupby('Coffee_Type')['Sales'].mean()

# Line plot
plt.plot(avg_sales.index, avg_sales.values, marker='o', color='orange')
plt.title("Average Sales by Coffee Type")
plt.xlabel("Coffee Type")
plt.ylabel("Average Sales")
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

#### Insights:
 
1.1st class paid the highest fare, 3rd class the least.
 
2.Higher fare correlates with higher survival chances.

### 8. Visualization with Seaborn

```python
## Histogram for coffee sales 
sns.histplot(df['Sales'], bins=15, kde=True, color='sienna')
plt.title("Distribution of Coffee Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

### Insights:
 
1.Most passengers were children/young adults, peak around 20-30 years.

```python
# Count plot for Coffee Type
sns.countplot(x='Coffee_Type', data=df, palette='Oranges_r')
plt.title("Number of Sales by Coffee Type")
plt.xlabel("Coffee Type")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### Insights:
 
1.Females survived more than males.
 
2.Confirms the "women and children first" pattern.

```python
# Box Plot for Sales by Coffee Type
plt.figure(figsize=(8,5))
sns.boxplot(x='Coffee_Type', y='Sales', data=df, palette='Oranges')
plt.title("Sales Distribution by Coffee Type")
plt.xlabel("Coffee Type")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### Insights:
 
1.1st class fares are more spread out, median higher.
 
2.3rd class has low fares with few outliers.
