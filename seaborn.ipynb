## seaborn- statistical Data visualization

#### 1.What is seaborn?
Seaborn is a python library that builds on the top of a matplotlib that makes the statistical data more easy and beautiful.

#### why we need it?
1. For creating box plots like violin pair plots.
2. automatically handles data from pandas dataframes.

#### where we use it?
1. Data analysis and reporting
2. Exploring relationships between variables
3. Machine learning for visualizing datasets and results.

```python
import matplotlib.pyplot as plt
## 4. Histograms & Countplots
# Histogram
sns.histplot(df['age'].dropna(), bins=10, kde=True)
plt.title("Histogram of Age")
plt.show()
 
# Countplot
sns.countplot(x='class', data=df)
plt.title("Count of Passengers by Class")
plt.show()
```

##### 5. Scatter plot Age vs fare

```python
sns.scatterplot(x='age', y='fare', data=df, hue='sex')
plt.title("Age vs Fare")
plt.show()
```

```python
### 6. Boxplot & Violin Plot
# Boxplot
sns.boxplot(x='class', y='age', data=df)
plt.title("Boxplot of Age by Class")
plt.show()
 
# Violin plot
sns.violinplot(x='class', y='age', hue='sex', data=df, split=True)
plt.title("Violin Plot of Age by Class and Sex")
plt.show()
```

```python
### 7. Pairplot
iris = sns.load_dataset('iris')
sns.pairplot(iris, hue='species')
plt.show()
```

```python
### 8. Heatmap
# Correlation heatmap (numeric columns only)
corr = iris.select_dtypes(include=['float64', 'int64']).corr()
 
import seaborn as sns
import matplotlib.pyplot as plt
 
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap of Iris Dataset")
plt.show()
```

```python
### 9. Customization & Themes
sns.set_theme(style="whitegrid")  # options: darkgrid, white, ticks, etc.
sns.histplot(df['age'].dropna(), bins=10, color='orange')
plt.title("Customized Histogram with Seaborn")
plt.show()
```

```python
#1.Compare Passenger Survival by Class and Gender
sns.countplot(x='class', hue='sex', data=df)
plt.title("Passenger Survival by Class and Gender")
plt.show()
```

```python
#2.⁠ ⁠Distribution of Passenger Ages
sns.histplot(data=df, x='age', bins=30, kde=True)
plt.title("Distribution of Passenger Ages on the Titanic")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
#3.⁠ ⁠Fare Distribution by Class
sns.boxplot(x='class', y='fare', data=df)
plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.show()
```

```python
#4.Relationship Between Age and Fare
sns.scatterplot(data=df, x='age', y='fare', alpha=0.7)
plt.title("Relationship Between Age and Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
#5.⁠ ⁠Correlation Heatmap
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
numeric_df = df.corr(numeric_only=True)
plt.figure(figsize=(10, 8))
sns.heatmap(numeric_df, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title("Correlation Heatmap of Titanic Dataset")
plt.show()
```
