### 2. Import Libraries

```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

### 3. Load Dataset

```python
df = pd.read_csv("disney_plus_shows.csv")
print("✅ Dataset loaded successfully!")
```

### 4. Explore Data

```python
print("Shape:", df.shape)
df.head()
df.info()
df.describe(include='all').T
```

### 5. Data Cleaning

```python
print("\nMissing Values:")
print(df.isnull().sum())
# Drop duplicates
df.drop_duplicates(inplace=True)

# Handle missing data (example: fill or drop)
df['rating'].fillna("Not Rated", inplace=True)
df['release_year'].fillna(df['release_year'].median(), inplace=True)
```

### Insights:
 
1.Missing values were filled or removed.

2.Duplicates were dropped.

3.Data was ready for visualization and statistical analysis.

### 6. Analysis with NumPy & Pandas

```python
# Number of movies vs TV shows
sns.countplot(x='type', data=df, palette='Set2')
plt.title("Movies vs TV Shows on Disney Plus")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()
 
# Top 10 genres
if 'genre' in df.columns:
    top_genres = df['genre'].value_counts().head(10)
    sns.barplot(x=top_genres.values, y=top_genres.index, palette='viridis')
    plt.title("Top 10 Genres on Disney Plus")
    plt.xlabel("Count")
    plt.ylabel("Genre")
    plt.show()


# Content added over time
if 'release_year' in df.columns:
    sns.histplot(df['release_year'], bins=20, kde=True, color='orange')
    plt.title("Content Release Trend over the Years")
    plt.xlabel("Release Year")
    plt.ylabel("Count")
    plt.show()

# numpy 
release_years = df['release_year'].dropna().values
print("Average release year:", np.mean(release_years))
print("Oldest release:", np.min(release_years))
print("Most recent release:", np.max(release_years))

```

### Insights:
 
1.Average age: ~29 years.
 
2.Overall survival: ~38% passengers survived.
 
3.By class: 1st class had the highest survival, 3rd class the lowest.
 
4.By sex: Females had higher survival than males.

### 7. Visualization with Matplotlib

```python
#Bar chart
top_genres = df['genre'].value_counts().head(10)
sns.barplot(x=top_genres.values, y=top_genres.index)
```

#### Insights:
 
1.Visual confirms 2nd class passengers survived more.
 
2.Helps understand social/economic impact on survival.

```python
#Line Plots
if 'genre' in df.columns:
    top3 = df['genre'].value_counts().head(3).index
    for g in top3:
        trend = df[df['genre'] == g]['release_year'].value_counts().sort_index()
        plt.plot(trend.index, trend.values, marker='o', label=g)
    plt.title("Trends of Top 3 Genres Over the Years")
    plt.xlabel("Release Year")
    plt.ylabel("Number of Titles")
    plt.legend()
    plt.grid(True)
    plt.show()
```

#### Insights:
 
1.1st class paid the highest fare, 3rd class the least.
 
2.Higher fare correlates with higher survival chances.
